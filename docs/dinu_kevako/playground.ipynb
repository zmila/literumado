```python
import numpy as np
import matplotlib.pyplot as plt


def bezier_cubic(t, P0, P1, P2, P3):
    u = 1 - t
    return (
            u**3 * np.array(P0)
            + 3 * u**2 * t * np.array(P1)
            + 3 * u * t**2 * np.array(P2)
            + t**3 * np.array(P3)
    )


# fig, ax = plt.subplots(figsize=(8, 5))
fig, ax = plt.subplots(figsize=(16, 10))

# Оси
ax.axhline(0, color="black", linewidth=1)
ax.axvline(0, color="black", linewidth=1)

# Прямоугольник
rect = plt.Rectangle((-2, -1), 4, 2, fill=False, edgecolor="gray", linewidth=1.5)
ax.add_patch(rect)

# # Полуокружности
# theta_left = np.linspace(3*np.pi/2, np.pi/2, 150)
# ax.plot(-1 + np.cos(theta_left), np.sin(theta_left), 'blue', linewidth=2)

# theta_right = np.linspace(np.pi/2, -np.pi/2, 150)
# ax.plot(1 + np.cos(theta_right), np.sin(theta_right), 'red', linewidth=2)

# 3. Две окружности радиусом 1 с центрами (-1, 0) и (1, 0)
theta = np.linspace(0, 2 * np.pi, 300)

# Окружность 1: центр (-1, 0)
circle1_x = -1 + np.cos(theta)
circle1_y = np.sin(theta)
ax.plot(circle1_x, circle1_y, color="blue", linewidth=1.5)

# Окружность 2: центр (1, 0)
circle2_x = 1 + np.cos(theta)
circle2_y = np.sin(theta)
ax.plot(circle2_x, circle2_y, color="red", linewidth=1.5)


# Зелёная Безье
P0 = np.array([-1, -1])
P3 = np.array([1, 1])

# ПОДБИРАЙТЕ ЭТИ ЗНАЧЕНИЯ
a = 0.7
b = -0.8

P1 = np.array([a, b])
P2 = -P1  # автоматическое выполнение P1+P2=0

t_vals = np.linspace(0, 1, 200)
curve = np.array([bezier_cubic(t, P0, P1, P2, P3) for t in t_vals])
ax.plot(curve[:, 0], curve[:, 1], "green", linewidth=2.5, label=f"Безье P1=({a},{b})")

# Контрольные точки
ax.scatter([P1[0], P2[0]], [P1[1], P2[1]], color="lime", s=60, marker="x", linewidths=2)
ax.scatter([0], [0], color="green", s=80, edgecolors="black", zorder=5)
ax.scatter([-1, 1], [-1, 1], color="green", s=50, alpha=0.7)

# Точки на полуокружностях
ax.scatter([-1, -2, -1], [-1, 0, 1], color="blue", s=40, alpha=0.7)
ax.scatter([1, 2, 1], [1, 0, -1], color="red", s=40, alpha=0.7)

ax.set_aspect("equal")
ax.set_xlim(-2.5, 2.5)
ax.set_ylim(-1.5, 1.5)
ax.grid(True, linestyle=":", alpha=0.4)
ax.legend()
plt.show()
```
